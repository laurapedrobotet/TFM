```python
invalid_categories = {}

for i in range(1, 11):
    col_name = f'categoria_{i}'
    if col_name in users_clasificados.columns:
        unique_categories = users_clasificados[col_name].dropna().unique()

        for category in unique_categories:
            if category not in CATEGORIAS_C:
                if col_name not in invalid_categories:
                    invalid_categories[col_name] = []
                invalid_categories[col_name].append(category)

if invalid_categories:
    print("Se encontraron las siguientes categorías inválidas:")
    for col, categories in invalid_categories.items():
        print(f"  Columna '{col}': {', '.join(categories)}")
else:
    print("Todas las categorías en las columnas 'categoria_N' son válidas y están presentes en CATEGORIAS_C.")
```

```python
import pandas as pd

all_categories = pd.Series(dtype='object')

for i in range(1, 11):
    col_name = f'categoria_{i}'
    if col_name in users_clasificados.columns:
        all_categories = pd.concat([all_categories, users_clasificados[col_name].dropna()])

category_counts = all_categories.value_counts()

print("Frecuencia de cada categoría en el DataFrame 'users_clasificados':")
display(category_counts)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 10))
plt.pie(category_counts, labels=category_counts.index, autopct='%1.1f%%')
plt.title('Distribución de Categorías en users_clasificados (Todas las categorías)')
plt.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```

```python
categoria_cols = [f"categoria_{i}" for i in range(1, 11)]

df_categorias = pd.DataFrame({
    "user_id": users_clasificados["user_id"],
    "categorias": users_clasificados[categoria_cols].apply(
        lambda row: [x for x in row if pd.notna(x)],
        axis=1
    )
})

df_categorias.head()
```

```python
df_categorias["n_categorias"] = df_categorias["categorias"].apply(len)

df_categorias["n_categorias"].describe()
```

## **Top 10 categorías de interés más frecuentes**

```python
df_categorias = pd.read_csv(
    "/content/drive/MyDrive/TFM/8. Categorizacion/categorias_por_usuario_prompt2.csv")
```

```python
from collections import Counter

categorias_exploded = df_categorias.explode("categorias")

# Conteo
top10 = (
    categorias_exploded["categorias"]
    .value_counts()
    .head(10)
    .reset_index()
)

top10.columns = ["categoria", "n_usuarios"]

top10
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))

colors = plt.cm.tab10(range(len(top10)))
plt.barh(
    top10["categoria"][::-1],
    top10["n_usuarios"][::-1],
    color=colors
)

plt.xlabel('Número de usuarios', fontsize=12, fontweight='bold')
plt.ylabel('Categoría de interés', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.savefig("grafico_top10.png", dpi=300, bbox_inches="tight")
plt.show()
```
